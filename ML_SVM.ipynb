```python
import pandas as pd
import numpy as np
import matplotlib as lib
import seaborn as sns
import matplotlib.pyplot as plt
```

```python

from sklearn.datasets import load_breast_cancer
cancer= load_breast_cancer()
cancer.keys()
print(cancer['DESCR'])
```

```python
#data observation
cancer['data']
cancer['feature_names']
# cancer.keys()
# cancer['filename']
# cancer['target']
# cancer['data_module']
```

```python
df_feature=pd.DataFrame(cancer['data'], columns=cancer['feature_names'])
```

```python
cancer['target']
```

```python
cancer['target_names']
```

```python
df_Target= pd.DataFrame(cancer['target'],columns=['is_maligant_cancer'])
# df_Target
df_Target.head(20)
```

```python
X= df_feature
y=df_Target
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.svm import SVC
```

## Model Building

```python


model= SVC()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
y_pred
```

## Model evaluation

```python

from sklearn.metrics import classification_report
print(classification_report(y_pred,y_test))
# print(classification_report(y_test,y_pred))
```

## Optimizing hyperparameters
1.Kernel
2.C

## RUN VSM with "linear", C ="10"

```python
from  sklearn.metrics import accuracy_score

linear_svc= SVC(kernel="linear",C=10.0)
linear_svc.fit(X_train,y_train)
y_pred_svc= linear_svc.predict(X_test)
print(f"Linear Model Accuracy:, {accuracy_score( y_test,y_pred_svc)}")
```

```python
linear_svc.score(X_train,y_train)
```

```python
linear_svc= SVC(kernel="rbf",C=10.0)
linear_svc.fit(X_train,y_train)
y_pred_svc= linear_svc.predict(X_test)
print(f"Linear Model Accuracy:, {accuracy_score( y_test,y_pred_svc)}")
```

```python
linear_svc= SVC(kernel="sigmoid",C=10)
linear_svc.fit(X_train,y_train)
y_pred_svc= linear_svc.predict(X_test)
print(f"Linear Model Accuracy:, {accuracy_score( y_test,y_pred_svc)}")
```

```python
linear_svc= SVC(kernel="poly",C=60)
linear_svc.fit(X_train,y_train)
y_pred_svc= linear_svc.predict(X_test)
print(f"Linear Model Accuracy:, {accuracy_score( y_test,y_pred_svc)}")
```

```python
from sklearn.model_selection import GridSearchCV

param_grid={
    "C":[1,2,3,4,5,6],
    'kernel':['rbf','linear']
}

opt_model= GridSearchCV(SVC(),param_grid)
opt_model.fit(X_train,y_train)
y_pred_opt = opt_model.predict(X_test)
y_pred_opt

```

```python
opt_model.best_params_
```

```python
opt_model.score(X_train,y_train)
```

```python
opt_model.score(X_test,y_test)
```
